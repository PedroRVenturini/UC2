```python
import pandas as pd
import numpy as np
```

```python

df_2024_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.01.csv', sep=';', decimal=',', low_memory=False)
df_2024_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.02.csv', sep=';', decimal=',', low_memory=False)
df_2025_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.01.csv', sep=';', decimal=',', low_memory=False)
df_2025_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.02.csv', sep=';', decimal=',', low_memory=False)
```

```python
df = pd.concat([df_2024_1, df_2024_2, df_2025_1, df_2025_2], ignore_index=True)
```

```python
df_gasolina_rj = df[(df['Estado - Sigla'] == 'RJ') & (df['Produto'] == 'GASOLINA')]
```

```python
df_gasolina_rj
```

```python
media_municipio = df_gasolina_rj.groupby('Municipio')['Valor de Venda'].mean().reset_index()

media_municipio
```

```python
media_municipio = media_municipio.sort_values(by='Valor de Venda', ascending=False)
media_municipio
```

```python
array_gasolina = np.array(df_gasolina_rj['Valor de Venda'])

q1 = np.percentile(array_gasolina, 25)
q3 = np.percentile(array_gasolina, 75)

print(q1)
print(q3)
```

```python
iqr = q3 - q1

print(iqr)
```

```python
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

print(limite_inferior)
print(limite_superior)
```

```python
outliers = df_gasolina_rj[
    (df_gasolina_rj['Valor de Venda'] < limite_inferior) |
    (df_gasolina_rj['Valor de Venda'] > limite_superior)
]

outliers
```

```python
outliers_baixos = outliers[outliers['Valor de Venda'] < limite_inferior]

outliers_altos = outliers[outliers['Valor de Venda'] > limite_superior]

print('Outliers baixos:', len(outliers_baixos))
print('Outliers altos:', len(outliers_altos))
print('Total:', len(outliers))
```

```python
postos_outliers = outliers[
    ['Municipio', 'Revenda', 'Bairro', 'Valor de Venda']
]

postos_outliers
```

```python
df_etanol_rj = df[(df['Estado - Sigla'] == 'RJ') & (df['Produto'] == 'ETANOL')]

df_etanol_sp = df[(df['Estado - Sigla'] == 'SP') & (df['Produto'] == 'ETANOL')]

print('RJ:', len(df_etanol_rj))
print('SP:', len(df_etanol_sp))
```

```python
media_etanol_rj = df_etanol_rj['Valor de Venda'].mean()
media_etanol_sp = df_etanol_sp['Valor de Venda'].mean()

print('Média RJ:', media_etanol_rj)
print('Média SP:', media_etanol_sp)
```

```python
array_etanol_rj = np.array(df_etanol_rj['Valor de Venda'])

q1_rj = np.percentile(array_etanol_rj, 25)
q3_rj = np.percentile(array_etanol_rj, 75)

print('Q1 RJ:', q1_rj)
print('Q3 RJ:', q3_rj)
```

```python
array_etanol_sp = np.array(df_etanol_sp['Valor de Venda'])

q1_sp = np.percentile(array_etanol_sp, 25)
q3_sp = np.percentile(array_etanol_sp, 75)

print('Q1 SP:', q1_sp)
print('Q3 SP:', q3_sp)
```

```python
iqr_rj = q3_rj - q1_rj
iqr_sp = q3_sp - q1_sp

print('IQR RJ:', iqr_rj)
print('IQR SP:', iqr_sp)
```

```python
limite_inferior_rj = q1_rj - 1.5 * iqr_rj
limite_superior_rj = q3_rj + 1.5 * iqr_rj

limite_inferior_sp = q1_sp - 1.5 * iqr_sp
limite_superior_sp = q3_sp + 1.5 * iqr_sp

print('Limite inferior RJ:', limite_inferior_rj)
print('Limite superior RJ:', limite_superior_rj)

print('Limite inferior SP:', limite_inferior_sp)
print('Limite superior SP:', limite_superior_sp)
```

```python
outliers_rj = df_etanol_rj[
    (df_etanol_rj['Valor de Venda'] < limite_inferior_rj) |
    (df_etanol_rj['Valor de Venda'] > limite_superior_rj)
]

outliers_sp = df_etanol_sp[
    (df_etanol_sp['Valor de Venda'] < limite_inferior_sp) |
    (df_etanol_sp['Valor de Venda'] > limite_superior_sp)
]

print('Outliers RJ:', len(outliers_rj))
print('Outliers SP:', len(outliers_sp))
```

```python
outliers_baixos_rj = outliers_rj[
    outliers_rj['Valor de Venda'] < limite_inferior_rj
]

outliers_altos_rj = outliers_rj[
    outliers_rj['Valor de Venda'] > limite_superior_rj
]

outliers_baixos_sp = outliers_sp[
    outliers_sp['Valor de Venda'] < limite_inferior_sp
]

outliers_altos_sp = outliers_sp[
    outliers_sp['Valor de Venda'] > limite_superior_sp
]

print('RJ baixos:', len(outliers_baixos_rj))
print('RJ altos:', len(outliers_altos_rj))

print('SP baixos:', len(outliers_baixos_sp))
print('SP altos:', len(outliers_altos_sp))
```

```python
df_gasolina_rio = df[
    (df['Estado - Sigla'] == 'RJ') &
    (df['Municipio'] == 'RIO DE JANEIRO') &
    (df['Produto'] == 'GASOLINA')
]

print(len(df_gasolina_rio))
```

```python
media_bairro = df_gasolina_rio.groupby('Bairro')['Valor de Venda'].mean().reset_index()

media_bairro = media_bairro.sort_values(
    by='Valor de Venda',
    ascending=False
)

media_bairro
```

```python
array_bairros = np.array(media_bairro['Valor de Venda'])

q1_bairros = np.percentile(array_bairros, 25)
q3_bairros = np.percentile(array_bairros, 75)

print('Q1:', q1_bairros)
print('Q3:', q3_bairros)
```

```python
iqr_bairros = q3_bairros - q1_bairros

print('IQR:', iqr_bairros)
```

```python
limite_inferior_bairros = q1_bairros - 1.5 * iqr_bairros
limite_superior_bairros = q3_bairros + 1.5 * iqr_bairros

print('Limite inferior:', limite_inferior_bairros)
print('Limite superior:', limite_superior_bairros)
```

```python
outliers_bairros = media_bairro[
    (media_bairro['Valor de Venda'] < limite_inferior_bairros) |
    (media_bairro['Valor de Venda'] > limite_superior_bairros)
]

outliers_bairros
```
