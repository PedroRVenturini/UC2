```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt


categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 
'Casa']
valores = [4200, 2800, 3600, 1900]

plt.bar(categorias, valores)
plt.title('Total de Vendas por Categoria — NaraStore')
plt.show()
```

```python
df = pd.read_csv('vendas_narastore.csv')
df
```

```python
df_cat = df.groupby('categoria')['total_venda'].sum().reset_index().sort_values(by='total_venda', ascending= False)

plt.bar(df_cat['categoria'], df_cat['total_venda'])
plt.title('Total de Vendas por Categoria — NaraStore')
plt.xticks(rotation=45)
plt.show()
```
